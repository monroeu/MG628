# Task 7: Learn the Individual Statistics Functions

Below are the most important statistics functions used in descriptive analytics.

| Statistic | Pandas Function | Meaning |
|---|---|---|
| Count | `.count()` | Number of non-missing values |
| Mean | `.mean()` | Average value |
| Median | `.median()` | Middle value when sorted |
| Mode | `.mode()` | Most frequently occurring value |
| Standard Deviation | `.std()` | How spread out the values are |
| Minimum | `.min()` | Smallest value |
| Maximum | `.max()` | Largest value |

Run the code below and compare the outputs.

````python
print("COUNT")
display(df[numeric_columns].count())

print("\nMEAN")
display(df[numeric_columns].mean())

print("\nMEDIAN")
display(df[numeric_columns].median())

print("\nMODE")
# mode() may return more than one row if there are multiple modes.
# iloc[0] selects the first mode for each column.
display(df[numeric_columns].mode().iloc[0])

print("\nSTANDARD DEVIATION")
display(df[numeric_columns].std())

print("\nMINIMUM")
display(df[numeric_columns].min())

print("\nMAXIMUM")
display(df[numeric_columns].max())
````

# Task 11: Group Data by Product

The `groupby()` function lets us summarize data by category.

For example, to calculate total revenue by product:

```python
df.groupby('Product')['Total Revenue'].sum()
```

This means:

1. Group the rows by product.
2. Select the Total Revenue column.
3. Add up total revenue for each product.

````python
# Calculate total revenue by product.
revenue_by_product = df.groupby('Product')['Total Revenue'].sum().sort_values(ascending=False)

# Display the result as a table.
display(revenue_by_product.to_frame(name='Total Revenue'))
````

# Task 12: Create a Bar Chart for Total Revenue by Product

A bar chart is useful for comparing categories.

In this chart:

- The x-axis shows the products.
- The y-axis shows total revenue.

````python
plt.figure(figsize=(10, 6))
plt.bar(revenue_by_product.index, revenue_by_product.values)
plt.xlabel('Product')
plt.ylabel('Total Revenue ($)')
plt.title('Practice Chart: Total Revenue by Product')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
````

# Task 16: Create a Histogram for Unit Price

A **histogram** shows the distribution of a numerical variable.

For Unit Price, a histogram helps us understand whether prices are:

# Task 18: Detect Outliers Using the IQR Method

The IQR method uses quartiles to define possible outliers.

Key terms:

- **Q1** = 25th percentile
- **Q3** = 75th percentile
- **IQR** = Q3 - Q1
- **Lower Bound** = Q1 - 1.5 × IQR
- **Upper Bound** = Q3 + 1.5 × IQR

Any value below the lower bound or above the upper bound is considered a possible outlier.

````python
def detect_outliers(dataframe, column):
    # Detect outliers in a column using the IQR method.
    Q1 = dataframe[column].quantile(0.25)
    Q3 = dataframe[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = dataframe[(dataframe[column] < lower_bound) | (dataframe[column] > upper_bound)]
    return Q1, Q3, IQR, lower_bound, upper_bound, outliers

# Detect outliers for Units Sold.
units_Q1, units_Q3, units_IQR, units_lower, units_upper, units_outliers = detect_outliers(df, 'Units Sold')

# Detect outliers for Total Revenue.
rev_Q1, rev_Q3, rev_IQR, rev_lower, rev_upper, rev_outliers = detect_outliers(df, 'Total Revenue')

print("Units Sold Outlier Results")
print(f"Q1: {units_Q1:.2f}")
print(f"Q3: {units_Q3:.2f}")
print(f"IQR: {units_IQR:.2f}")
print(f"Lower Bound: {units_lower:.2f}")
print(f"Upper Bound: {units_upper:.2f}")
print(f"Number of Outliers: {len(units_outliers)}")
display(units_outliers)

print("\nTotal Revenue Outlier Results")
print(f"Q1: {rev_Q1:.2f}")
print(f"Q3: {rev_Q3:.2f}")
print(f"IQR: {rev_IQR:.2f}")
print(f"Lower Bound: {rev_lower:.2f}")
print(f"Upper Bound: {rev_upper:.2f}")
print(f"Number of Outliers: {len(rev_outliers)}")
display(rev_outliers)
````
